```python
import requests
from datetime import datetime, timezone

token_id = "63842529068710005716169325380315470359047749786610778647370693404952498013178"

start_ts = int(datetime(2026, 8, 28, 0, 0, tzinfo=timezone.utc).timestamp())
end_ts   = int(datetime(2026, 8, 29, 0, 0, tzinfo=timezone.utc).timestamp())

r = requests.get(
    "https://data-api.polymarket.com/v2/prices-history",
    params={
        "tokenId": token_id,
        "start": start_ts,
        "end": end_ts,
        "bucketSeconds": 60,
    },
    timeout=20,
)
print(r.status_code)
print(r.json())
```

```python
for bucket in [60, 3600, 43200]:  # 1min, 1hr, 12hr
    r = requests.get(
        "https://data-api.polymarket.com/v2/prices-history",
        params={
            "tokenId": token_id,
            "start": start_ts,
            "end": end_ts,
            "bucketSeconds": bucket,
        },
        timeout=20,
    )
    data = r.json().get("data", [])
    print(f"bucketSeconds={bucket}: {len(data)} points")
```
